## What is an LLM? 

The Core Definition: An LLM is a probabilistic model trained to predict the next word (token) in a sequence.
The Mechanism: It doesn't "know" facts; it calculates the likelihood of a word appearing given the context.
The Scale:
    - Small: "The cat sat on the..." $\rightarrow$ "mat" (easy).
    - Large: "The derivative of $x^2$ is..." $\rightarrow$ "$2x$" (requires math knowledge).
    - Huge: "Write a Python script to..." $\rightarrow$ (requires reasoning).

## How do Machines "Read"? (Tokenization)

- Computers don't see text. They only see numbers.
- Tokenization: We chop text into chunks called Tokens.
- Sometimes a token is a word ("apple"), sometimes part of a word ("ing").
- The Map: We use a massive dictionary (Vocabulary) to turn these tokens into integers.
    - "Hello" $\rightarrow$ 15496
    - "World" $\rightarrow$ 9231

## The Old Way vs. New Way

RNNs (Recurrent Neural Networks): Read one word at a time. Forgot the beginning of the sentence by the time they reached the end. (Sequential = Slow).

Transformers (2017):

Parallelism: Read the whole sentence at once.

Self-Attention: The breakthrough mechanism. It allows the model to look at all words simultaneously and decide which ones relate to each other.

## Training

**Pre-training (The Library):**

Goal: Learn grammar, facts, and reasoning.

Data: The entire internet (CommonCrawl, Wikipedia, GitHub).

Task: "Predict the next token."

Result: A smart but chaotic simulator (Base Model).

**Fine-Tuning (The Textbooks):**

Goal: Learn to follow instructions (Q&A, Summarization).

Data: High-quality Q&A pairs written by humans.

Result: A helpful assistant (Instruct Model).

**RLHF (The Etiquette School):**

Goal: Safety and preference (Reinforcement Learning from Human Feedback).

Mechanism: Humans rank outputs (A is better than B). The model learns to align with human values.

## Why do they hallucinate?

The Feature, not a Bug: LLMs are Probabilistic, not deterministic.

Temperature: A setting that controls creativity.

Temperature 0.0 = Always pick the most likely word (Rigid, factual).

Temperature 1.0 = Pick risky words (Creative, prone to lying).

Source Amnesia: The model remembers the pattern of words (e.g., "The capital of France is Paris") but forgets where it learned it. If the training data contained misconceptions, the model treats them as fact.

## Model Parameters

- There are few things to know on the different options you have for using openai models
- The first and most important thing is the model: https://platform.openai.com/docs/models/overview
- Then there are some parameters that are important:

`messages`
A list of messages comprising the conversation so far. Can also include a python function

```python
messages=[
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Knock knock."},
    {"role": "assistant", "content": "Who's there?"},
    {"role": "user", "content": "Orange."},
]
```

- Messages usually come with `content` and a `role`
- Roles:
    - system
    - user
    - assistant
    - function (might discuss this)


`frequency_penalty`/`presence penalty`
Defaults to 0
Number between -2.0 and 2.0. Positive values penalize new tokens based on their existing frequency in the text so far, decreasing the model's likelihood to repeat the same line verbatim.

Frequency_penalty: This parameter is used to discourage the model from repeating the same words or phrases too frequently within the generated text. It is a value that is added to the log-probability of a token each time it occurs in the generated text. A higher frequency_penalty value will result in the model being more conservative in its use of repeated tokens.

Presence_penalty: This parameter is used to encourage the model to include a diverse range of tokens in the generated text. It is a value that is subtracted from the log-probability of a token each time it is generated. A higher presence_penalty value will result in the model being more likely to generate tokens that have not yet been included in the generated text.

Example: 

You ask: "What is the temperature today?"

Low presence penalty: "The temperature today is 75 degrees."

High presence penalty: "On this morrow, the amount of warmth is at 75 degrees."

Low frequency penalty: "The temperature today is a temperature that is the temperature today, 75 degrees."

High frequency penalty: "It is 75 degrees."


`logit_bias`

Accepts a json object that maps tokens (specified by their token ID in the tokenizer) to an associated bias value from -100 to 100. Mathematically, the bias is added to the logits generated by the model prior to sampling. The exact effect will vary per model, but values between -1 and 1 should decrease or increase likelihood of selection; values like -100 or 100 should result in a ban or exclusive selection of the relevant token.

`max_tokens`

The maximum number of tokens to generate in the chat completion.

The total length of input tokens and generated tokens is limited by the model's context length. 

`n`

How many chat completion choices to generate for each input message.



`stop`
string / array / null
Optional
Defaults to null
Up to 4 sequences where the API will stop generating further tokens.

`stream`
boolean or null
Optional
Defaults to false
If set, partial message deltas will be sent, like in ChatGPT. Tokens will be sent as data-only server-sent events as they become available, with the stream terminated by a data: [DONE] message. Example Python code.

**`temperature`**
number or null
Optional
Defaults to 1
What sampling temperature to use, between 0 and 2. Higher values like 0.8 will make the output more random, while lower values like 0.2 will make it more focused and deterministic.

We generally recommend altering this or top_p but not both.

**`top_p`**
number or null
Optional
Defaults to 1
An alternative to sampling with temperature, called nucleus sampling, where the model considers the results of the tokens with top_p probability mass. So 0.1 means only the tokens comprising the top 10% probability mass are considered.

We generally recommend altering this or temperature but not both.

`user`
string
Optional
A unique identifier representing your end-user, which can help OpenAI to monitor and detect abuse. Learn more.

## Prompt Engineering

- Writing prompts are just as important as understanding the code here
- The precision of your prompt compensates for the fact that you don't need to write code or train a model
- You need to write a prompt as precisely and clearly as possible so the model gives you what you actually want
- These techniques apply across providers (OpenAI, Anthropic, Google) — the underlying model may differ but the principles are universal

1. Tell the model what you want at the beginning AND end
    - Before going into an explanation, the prompt should specify the task at the very beginning.
    - Then it should repeat the task at the end again.
    - This works because attention is strongest at the start and end of the context window.

```
Your task is to verify if the statement "Several sources mention a chance of another large eruption" is supported by a specific quote from the following set of snippets.
---
SNIPPETS
[1] 14 percent chance of megaquake hitting Seattle, experts say
SEATTLE - There's a 14 percent chance of a magnitude 9 Cascadia earthquake hitting Seattle in the next 50 years, the U.S. Geological Survey estimates. "Unfortunately, we are unable to...

[2] Earthquake experts lay out latest outlook for Seattle's 'Really Big One'
"We say that there's approximately a 14% chance of another approximately magnitude-9 earthquake occurring in the next 50 years," said a geophysicist at the University of Washington...
---
Is the statement "Several sources mention a chance of another large eruption" directly implied or stated by the snippets?
```

2. Prime the output
    - This is critical. You often want the output of the model to NOT be a conversation, but something in a data format. Priming the prompt which includes information in the output you want it in, makes it more likely that the model will output in that form.

```
John Smith is married to Lucy Smith. They have five kids, and he works as a software engineer at Microsoft. What search queries should I do to fact-check this? ## One possible search query is:

1. How many kids do they have?
2. What is Lucy's maiden name?

```

or

```
Here’s a bulleted list of key points:\n-
```


3. Using clear syntax and delineating different parts of the prompt
    - Use separators (`---`, XML tags, or labeled sections) to distinguish instructions from data.

```
You will read a paragraph, and then issue queries to a search engine in order to fact-check it. Also explain the queries.
---
PARAGRAPH
John Smith is married to Lucy Smith. They have five kids, and he works as a software engineer at Microsoft. What search queries should I do to fact-check this?
---
QUERIES
```

   - XML-style tags are especially effective because they make boundaries unambiguous:

```
You are a research assistant. Analyze the following job posting.

<job_posting>
We are hiring a Data Analyst. Flexible start times. Bachelor's degree required.
</job_posting>

Extract: is_flexible_schedule (0 or 1), education_required (High School/BA/MA/PhD).
```

4. Break the job down into smaller tasks
    - If you need the model to do several things, separating them into steps (or separate API calls) is more reliable than asking for everything at once.
    - This is the foundation of the multi-agent systems we saw earlier.

5. Chain of Thought Prompting
    - Ask the model to break down its reasoning before giving a final answer.
    - This dramatically improves accuracy on multi-step problems — the model "shows its work."

```
Q:
Who was the most decorated (maximum medals) individual athlete in the Olympic games that were held at Sydney? Take a step-by-step approach in your response, cite sources and give reasoning before sharing final answer in the below format: ANSWER is: <name>
```

```
A:
Step 1: Determine the year of the Sydney Olympic games. The Sydney Olympic games were held in the year 2000.

Step 2: Find a list of individual athletes who won multiple medals in the 2000 Sydney Olympics.

According to the official website of the Olympic Games, there were several individual athletes who won multiple medals in the 2000 Sydney Olympics. Some of the notable names include:
• Ian Thorpe (Australia) - 3 gold, 2 silver, 0 bronze
• Pieter van den Hoogenband (Netherlands) - 2 gold, 0 silver, 2 bronze
• Lenny Krayzelburg (USA) - 3 gold, 0 silver, 0 bronze
• Inge de Bruijn (Netherlands) - 3 gold, 1 silver, 0 bronze

Step 3: Determine the athlete with the most medals. After reviewing the list of individual athletes who won multiple medals in the 2000 Sydney Olympics, we can see that the athlete with the most medals is Ian Thorpe from Australia, who won a total of 5 medals (3 gold and 2 silver).

Therefore, the ANSWER is: Ian Thorpe.
```

   - Newer APIs expose this explicitly: OpenAI's `o3`/`o4` reasoning models and Anthropic's "extended thinking" run chain-of-thought internally before producing output.

6. Controlling Hallucinations
    - Models "hallucinate": they generate plausible-sounding but false content.
    - Simply saying "only give true statements" is not enough.
    - Better strategies:
        - Ask for **citations** (or inline citations) so you can verify
        - **Ground the model** in a provided context (tip 7) rather than asking it to recall facts from training
        - Use **structured output** (tip below) so the model can only produce well-formed data, not freeform text
        - Set `temperature` low (0.0–0.2) for factual tasks

7. Grounding with context
    - If you need analysis, not creativity, provide the data directly in the prompt.
    - The model draws answers from your text rather than from potentially-wrong training data.
    - This is the core idea behind Retrieval-Augmented Generation (RAG).

8. Giving Examples (Few-Shot Prompting)
    - Showing the model one or two examples of the exact input/output format you want is often the single most effective technique.
    - You can do this by including prior turns in the `messages` list:

```python
messages=[
    {"role": "system",    "content": "You are a helpful assistant."},
    {"role": "user",      "content": "Tell me all the vowels in this sentence."},
    {"role": "assistant", "content": "There are 4 vowels in this sentence."},
    {"role": "user",      "content": "How many vowels are in this sentence now?"},
]
```

## Getting it out as data

Specifying the output format is one of the most important prompt engineering decisions.

**The old way (fragile):** Ask the model to "output JSON" in the prompt. It mostly complies, but sometimes wraps the JSON in ```json ... ``` fences, adds an explanation, or omits a field. You end up writing defensive parsing code.

**The modern way (reliable):** Use `response_format={"type": "json_object"}` in the API call. The model is constrained at the token level to only emit valid JSON — no fences, no commentary. Combine this with a schema description in the system prompt.

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "Extract variables. Output ONLY valid JSON with keys: flexible_schedule (int), education (str)."},
        {"role": "user",   "content": f"Job description: {text}"},
    ],
    response_format={"type": "json_object"},
)
result = json.loads(response.choices[0].message.content)
```

For even stronger guarantees, the OpenAI API also supports **Structured Outputs** (`response_format={"type": "json_schema", "json_schema": {...}}`), where you supply a full JSON Schema and the model is forced to match it exactly — useful when you need every field and every type to be correct.

### Example: Classifying Household Coping Strategies after a Flood Shock

A standard task in development economics is coding open-ended survey responses about how households cope with shocks (floods, droughts, illness, job loss). This feeds into indices like the **WFP Reduced Coping Strategies Index (rCSI)**, which is used to measure food insecurity and resilience.

The problem: households describe coping in their own words. An RA reading 5,000 responses will be slow, expensive, and inconsistent. Regex fails immediately — "we ate less" and "we survived on wild fruits" are both food reduction but share no keywords.

Why a **base model struggles**: the categories require domain knowledge (knowing that selling a cow is an asset sale, not income), and responses are often colloquial, incomplete, or multilingual. Fine-tuning on RA-coded examples teaches the model the exact taxonomy your project uses.

**Coping strategy categories (simplified rCSI):**

| Code | Strategy | Example |
|---|---|---|
| `food_reduction` | Eating less / cheaper food / fewer meals | "We ate once a day for two weeks" |
| `asset_sale` | Selling livestock, tools, or household items | "We sold two goats to buy food" |
| `labor_migration` | Household member migrates or takes casual work | "My husband went to Kampala to find work" |
| `social_network` | Borrowing from or relying on family/community | "We stayed at my brother's and he fed us" |
| `formal_aid` | Government, NGO, or international assistance | "The Red Cross gave us food rations" |
| `no_coping` | Did nothing / unable to cope | "We just waited for the water to go down" |

````python
def validate_gpt(dataset):
    # Format error checks
    format_errors = defaultdict(int)

    for ex in dataset:
        if not isinstance(ex, dict):
            format_errors["data_type"] += 1
            continue

        messages = ex.get("messages", None)
        if not messages:
            format_errors["missing_messages_list"] += 1
            continue

        for message in messages:
            if "role" not in message or "content" not in message:
                format_errors["message_missing_key"] += 1

            if any(
                k not in ("role", "content", "name", "function_call") for k in message
            ):
                format_errors["message_unrecognized_key"] += 1

            if message.get("role", None) not in (
                "system",
                "user",
                "assistant",
                "function",
            ):
                format_errors["unrecognized_role"] += 1

            content = message.get("content", None)
            function_call = message.get("function_call", None)

            if (not content and not function_call) or not isinstance(content, str):
                format_errors["missing_content"] += 1

        if not any(message.get("role", None) == "assistant" for message in messages):
            format_errors["example_missing_assistant_message"] += 1

    if format_errors:
        print("Found errors:")
        for k, v in format_errors.items():
            print(f"{k}: {v}")
    else:
        print("No errors found")
````

## Synthetic Data Generation: Use a Smart Model to Train a Cheap One

28 hand-coded examples is not enough for reliable fine-tuning — OpenAI recommends at least 50–100, and more is better.

But getting an RA to code hundreds of examples is expensive. The solution: **use a powerful model to generate synthetic training data, then fine-tune a smaller model on it**.

```
GPT-4o (smart, expensive)
    │
    │  generates 200 realistic labeled examples
    ▼
synthetic_training.jsonl
    │
    │  fine-tune
    ▼
gpt-4o-mini (fast, cheap)  ←── deploy this against 5,000 real survey responses
```

This works because:
- GPT-4o understands the task well enough to generate plausible, varied examples
- The fine-tuned GPT-4o-mini learns the specific taxonomy and output format
- At inference time you pay GPT-4o-mini prices, not GPT-4o prices
- The bottleneck shifts from RA time to prompt design

The key risk: if GPT-4o generates systematically wrong or biased examples, the fine-tuned model inherits those errors. Always validate a sample of synthetic data against real RA-coded responses before training.

````python
import json, time

CATEGORIES = {
    "food_reduction": "The household ate less food, reduced meal frequency, ate cheaper or lower-quality food, or went without meals.",
    "asset_sale": "The household sold livestock, tools, land, household items, or other assets to cope.",
    "labor_migration": "A household member migrated for work or took on casual/day labor outside the home.",
    "social_network": "The household borrowed from, relied on, or received help from family, friends, neighbors, or community/religious groups (not formal aid).",
    "formal_aid": "The household received food, cash, or goods from government programs, NGOs, or international organizations.",
    "no_coping": "The household did nothing, had no options, or only describes the shock without any active coping response.",
}

GENERATOR_SYSTEM = """
You are a survey data generator for a development economics research project in East/Central Africa.
Generate realistic, varied open-ended survey responses as a household member would give them verbally.
Responses should:
- Sound natural and colloquial, as if spoken by a rural household member
- Vary in length (5–30 words), specificity, and phrasing
- Reference locally plausible details (crops: maize, cassava, sorghum; livestock: goats, chickens, cows;
  places: Kampala, Jinja, the fishing village, town; currency: shillings; food: posho, chapati, ugali)
- Sometimes be ambiguous or brief, as real survey responses often are
Output a JSON array of strings. No labels, no explanations — only the responses.
"""


def generate_synthetic(category: str, description: str, n: int = 10) -> list[str]:
    """Ask GPT-4o to generate n synthetic survey responses for a given coping category."""
    response = client.chat.completions.create(
        model="gpt-4o",  # smart model — understands context and generates varied examples
        messages=[
            {"role": "system", "content": GENERATOR_SYSTEM},
            {
                "role": "user",
                "content": (
                    f"Generate {n} survey responses where the household's PRIMARY coping strategy is:\n"
                    f"Category: {category}\n"
                    f"Definition: {description}\n\n"
                    f"The question asked was: 'What did your household do to cope after the flood?'\n"
                    f"Return a JSON array of {n} strings."
                ),
            },
        ],
        temperature=0.9,  # high temperature for variety
        response_format={"type": "json_object"},
    )
    raw = json.loads(response.choices[0].message.content)
    # GPT-4o may wrap the array in a key — extract the first list value found
    if isinstance(raw, list):
        return raw
    elif isinstance(raw, dict):
        for v in raw.values():
            if isinstance(v, list):
                return v


# Generate 15 synthetic examples per category (90 total)
# In a real project you'd generate 50–200 per category
synthetic_responses, synthetic_codes = [], []

for category, description in CATEGORIES.items():
    print(f"Generating for: {category}...")
    examples = generate_synthetic(category, description, n=15)
    synthetic_responses.extend(examples[:15])
    synthetic_codes.extend([category] * 15)
    time.sleep(1)  # light rate-limit courtesy pause

print(f"\nSynthetic examples generated: {len(synthetic_responses)}")
print("\nSample synthetic responses:")
for resp, code in list(zip(synthetic_responses, synthetic_codes))[::15]:
    print(f"  [{code}] {resp}")
````

## Agentic AI

So far we have been using LLMs in a **request-response** pattern: you send a prompt, you get back text.

**Agentic AI** is a different paradigm:

> An agent is an LLM that can *decide what to do next*, take actions in the world, observe the results, and loop until the task is complete.

The key shift is **autonomy over multiple steps**. The model doesn't just answer — it plans, acts, checks its work, and adapts.

---

### Why does this matter for economists?

| Old workflow | Agentic workflow |
|---|---|
| You write code to call an API | The agent writes and runs the code |
| You clean data manually | The agent spots errors and fixes them |
| You search for papers by hand | The agent searches, reads, and summarizes |
| One prompt → one answer | One goal → many actions → final answer |

The bottleneck shifts from *doing the steps* to *specifying the goal clearly*.

## The Core Primitive: Tool Use

The main mechanism that makes agents work is **tool use** (also called *function calling*).

1. You define a set of **tools** — Python functions the model is allowed to call.
2. The model reads the task and decides *which tool to call* and *with what arguments*.
3. You execute the function and hand the result back to the model.
4. The model continues reasoning with the result.

```
User goal
   │
   ▼
Model thinks → decides to call tool_X(arg=...)
   │
   ▼
Your code runs tool_X, returns result
   │
   ▼
Model sees result → decides: done? or call another tool?
   │
   ▼
Final answer
```

The model never *runs* your code — it only produces a structured JSON request.  
**You** are responsible for executing it safely.

## Defining Tools

In the OpenAI API, tools are described as JSON schemas — name, description, and the parameters they accept.

The model reads the description to decide *when* to call a tool and *how* to fill the arguments.

OpenAI uses `finish_reason == "tool_calls"` to signal that the model wants to call one or more tools instead of returning a final answer.

````python
import requests

# ── Country name → ISO-2 code (World Bank uses ISO-2) ───────────────────────
COUNTRY_CODES = {
    "usa": "US",
    "us": "US",
    "united states": "US",
    "uk": "GB",
    "united kingdom": "GB",
    "britain": "GB",
    "germany": "DE",
    "france": "FR",
    "japan": "JP",
    "china": "CN",
    "india": "IN",
    "brazil": "BR",
    "kenya": "KE",
    "nigeria": "NG",
    "ethiopia": "ET",
    "mexico": "MX",
    "canada": "CA",
    "australia": "AU",
}


def get_inflation_rate(country: str, year: int) -> dict:
    """
    Fetch annual CPI inflation rate (%) from the World Bank API.
    Indicator FP.CPI.TOTL.ZG — no API key required.
    """
    iso2 = COUNTRY_CODES.get(country.lower().strip())
    if iso2 is None:
        # Try treating the input directly as an ISO-2 code
        iso2 = country.upper()

    url = (
        f"https://api.worldbank.org/v2/country/{iso2}"
        f"/indicator/FP.CPI.TOTL.ZG"
        f"?date={year}&format=json&per_page=1"
    )
    try:
        resp = requests.get(url, timeout=10)
        resp.raise_for_status()
        payload = resp.json()
        # World Bank returns [metadata, [datapoints]]
        records = payload[1]
        if not records or records[0]["value"] is None:
            return {"error": f"No World Bank data for {country} in {year}"}
        rate = round(records[0]["value"], 2)
        return {
            "country": country,
            "year": year,
            "inflation_rate_pct": rate,
            "source": "World Bank — FP.CPI.TOTL.ZG",
        }
    except Exception as e:
        return {"error": str(e)}


def compute_real_return(nominal_return_pct: float, inflation_rate_pct: float) -> dict:
    """Fisher equation: real return = (1 + nominal) / (1 + inflation) - 1"""
    real = ((1 + nominal_return_pct / 100) / (1 + inflation_rate_pct / 100) - 1) * 100
    return {"real_return_pct": round(real, 4)}


# ── Tool descriptions for the OpenAI API ──────────────────────────────────────
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_inflation_rate",
            "description": (
                "Returns the annual CPI inflation rate (%) for a given country and year "
                "using the World Bank API. Call this before computing real returns."
            ),
            "parameters": {
                "type": "object",
                "properties": {
                    "country": {
                        "type": "string",
                        "description": "Country name or ISO-2 code, e.g. 'USA' or 'US'",
                    },
                    "year": {
                        "type": "integer",
                        "description": "4-digit year, e.g. 2023",
                    },
                },
                "required": ["country", "year"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "compute_real_return",
            "description": "Converts a nominal return to a real (inflation-adjusted) return using the Fisher equation.",
            "parameters": {
                "type": "object",
                "properties": {
                    "nominal_return_pct": {
                        "type": "number",
                        "description": "Nominal return in percent",
                    },
                    "inflation_rate_pct": {
                        "type": "number",
                        "description": "Inflation rate in percent",
                    },
                },
                "required": ["nominal_return_pct", "inflation_rate_pct"],
            },
        },
    },
]

tool_registry = {
    "get_inflation_rate": get_inflation_rate,
    "compute_real_return": compute_real_return,
}

# Quick sanity check
print("Tools ready:", [t["function"]["name"] for t in tools])
print("Live test:", get_inflation_rate("USA", 2022))
````

## The Agent Loop

In the OpenAI API the loop looks like this:

1. Call `chat.completions.create()` with `tools=`.
2. If `finish_reason == "tool_calls"`: execute the requested functions, append the results as `role="tool"` messages, and call the model again.
3. If `finish_reason == "stop"`: the model is done — return its text.

````python
def run_agent(
    user_question: str,
    tools: list,
    system: str = "",
    max_iters: int = 10,
    verbose: bool = True,
) -> str:
    """
    A minimal OpenAI agent loop.
    Keeps calling the model until finish_reason == 'stop' or max_iters is reached.
    """
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": user_question})

    for iteration in range(max_iters):
        response = client.chat.completions.create(
            model="o3",
            tools=tools,
            messages=messages,
        )

        msg = response.choices[0].message
        finish = response.choices[0].finish_reason

        if verbose:
            print(f"Iteration {iteration + 1} — finish_reason: {finish}")

        # ── Model is done ─────────────────────────────────────────────────────
        if finish == "stop":
            return msg.content

        # ── Model wants to call tools ─────────────────────────────────────────
        elif finish == "tool_calls":
            # 1. Append the assistant's message (contains the tool_calls)
            messages.append(msg)

            # 2. Execute each requested call and append the result
            for tc in msg.tool_calls:
                fn = tool_registry[tc.function.name]
                args = json.loads(tc.function.arguments)
                result = fn(**args)

                if verbose:
                    print(f"  → {tc.function.name}({args}) = {result}")

                messages.append({
                    "role": "tool",
                    "tool_call_id": tc.id,
                    "content": json.dumps(result),
                })

        else:
            raise RuntimeError(f"Unexpected finish_reason: {finish}")

    return f"[Agent stopped after {max_iters} iterations]"


# ── Run it ────────────────────────────────────────────────────────────────────
answer = run_agent(
    "I earned a 7% nominal return on a US bond fund in 2022. What was my real return?",
    tools=tools,
    verbose=True,
)
print("\nFinal answer:", answer)
````

## A Richer Example: Automated Literature Search Agent

Suppose you are starting a research project and want to quickly build a reading list.  
A research agent can:

1. Search for relevant papers (tool: `search_papers`)
2. Fetch the abstract of a specific paper (tool: `get_abstract`)
3. Synthesize a concise summary with key citations

This is the kind of task that would take an RA an afternoon — an agent can do it in seconds.

````python
# Simulated paper database — in production, hit Semantic Scholar / arXiv APIs
PAPER_DB = [
    {
        "id": "P001",
        "title": "Minimum Wages and Employment: A Case Study of the Fast-Food Industry",
        "authors": "Card & Krueger",
        "year": 1994,
        "abstract": "We analyze the effect of a rise in NJ's minimum wage on fast-food employment. "
        "Contrary to standard theory, employment grew in NJ relative to PA after the increase.",
    },
    {
        "id": "P002",
        "title": "Using Differences in Knowledge Across Neighborhoods to Uncover the Impacts of the EITC",
        "authors": "Chetty, Friedman & Saez",
        "year": 2013,
        "abstract": "We study how knowledge of the EITC schedule varies across neighborhoods and use this variation "
        "to estimate the impact of EITC incentives on earnings.",
    },
    {
        "id": "P003",
        "title": "Causal Effects of Minimum Wages: New Evidence from a Border Discontinuity Design",
        "authors": "Dube, Lester & Reich",
        "year": 2010,
        "abstract": "We use policy discontinuities at state borders to estimate minimum wage effects, "
        "finding small disemployment effects and significant wage compression.",
    },
    {
        "id": "P004",
        "title": "The Labor Market Impacts of Youth Training: Experimental Evidence",
        "authors": "Card, Ibarrarán & Regalia",
        "year": 2011,
        "abstract": "A randomized evaluation of a youth training program finds "
        "significant earnings gains for women but not men.",
    },
]


def search_papers(query: str, max_results: int = 3) -> list:
    """Return papers whose title or abstract contains any query word."""
    query_words = set(query.lower().split())
    scored = []
    for p in PAPER_DB:
        text = (p["title"] + " " + p["abstract"]).lower()
        hits = sum(1 for w in query_words if w in text)
        if hits:
            scored.append((hits, p))
    scored.sort(key=lambda x: x[0], reverse=True)
    results = [p for _, p in scored[:max_results]]
    return results


def get_abstract(paper_id: str) -> dict:
    """Retrieve the full abstract for a paper by ID."""
    for p in PAPER_DB:
        if p["id"] == paper_id:
            return {"id": paper_id, "abstract": p["abstract"]}
    return {"error": f"Paper {paper_id} not found"}


research_tools = [
    {
        "type": "function",
        "function": {
            "name": "search_papers",
            "description": "Search an academic paper database by topic keywords. Returns title, authors, and year.",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "Keywords to search for",
                    },
                    "max_results": {
                        "type": "integer",
                        "description": "Maximum papers to return (default 3)",
                    },
                },
                "required": ["query"],
            },
        },
    },
    {
        "type": "function",
        "function": {
            "name": "get_abstract",
            "description": "Fetch the full abstract of a paper using its ID from a prior search.",
            "parameters": {
                "type": "object",
                "properties": {
                    "paper_id": {
                        "type": "string",
                        "description": "Paper ID from a search result",
                    },
                },
                "required": ["paper_id"],
            },
        },
    },
]

tool_registry.update({"search_papers": search_papers, "get_abstract": get_abstract})
````

## What Are the Research Tools, Actually?

`research_tools` is just a **schema** — a description sent to the model so it knows what tools exist and what arguments to provide. The model never sees or runs the Python functions. It only reads the JSON schema and produces a structured request like: "call `search_papers` with `query='minimum wage'`".

The actual execution happens in your Python code inside `run_agent`:

```python
for tc in msg.tool_calls:
    fn   = tool_registry[tc.function.name]   # look up the real Python function
    args = json.loads(tc.function.arguments) # parse the model's requested arguments
    result = fn(**args)                       # YOU run it, not the model
```

`tool_registry` is what connects the schema name (a string like `"search_papers"`) to the actual Python function. The model just outputs JSON; the agent loop is the glue that executes it.

| Thing | What it is |
|---|---|
| `research_tools` | Schema sent to the model — describes names, descriptions, and argument types |
| Model's `tool_calls` response | JSON saying which tool to call and with what args |
| `tool_registry` | Dict mapping tool names → real Python functions |
| `run_agent` loop | Your code that executes the functions and hands results back |

This is why the model never running your code matters: **you** control what actually executes, which means you are responsible for validating inputs and handling errors before calling `fn(**args)`.

## System Prompts for Agents

The **system prompt** steers the agent's behavior:
- What role it plays and what tools it has
- How to format output
- Constraints ("never make up citations")

A weak system prompt leads to hallucinations, wasted tool calls, and hard-to-debug loops.

````python
SYSTEM_PROMPT = """
You are a rigorous economic research assistant.

Your job:
1. Use search_papers to find literature on the user's topic.
2. Use get_abstract to read the full abstract of any promising paper.
3. Write a concise literature summary (3-5 sentences) citing papers as (Author, Year).

Rules:
- Never invent paper titles, authors, or results.
- If search returns nothing relevant, say so explicitly.
- Always read at least one abstract before writing your summary.
"""

answer = run_agent(
    "Summarize the literature on minimum wages and low-wage worker earnings.",
    tools=research_tools,
    system=SYSTEM_PROMPT,
    verbose=True,
)
print("\nFinal answer:")
print(answer)
````

## Best Practices and Risks

### Do
- **Validate tool inputs** before running them — the model can pass bad arguments.
- **Log every tool call and result** — agents are hard to debug without a full trace.
- **Set `max_iters`** — a buggy agent can loop forever.
- **Use the cheapest model that works** — agent loops multiply API costs.
- **Test tools in isolation** before connecting them to the agent.

### Don't
- Give an agent tools it doesn't need — more tools = more chances to pick the wrong one.
- Let an agent execute arbitrary code or hit production databases without human review.
- Trust the model's output as ground truth — always sanity-check results.

### The Control Problem in Research
Agents are powerful precisely because they act autonomously. In research, this creates risks:
- **Hallucinated citations** — the model may mix real papers with invented ones.
- **P-hacking by proxy** — an agent iterating over many models may surface spurious results.
- **Irreproducibility** — if the agent's actions aren't logged, the analysis can't be replicated.

The best agentic workflows keep a **human in the loop** at high-stakes decision points.

## Token Costs in Agent Loops

Every loop iteration sends the **full conversation history** to the model, so input token count grows **linearly** — each iteration adds roughly the same fixed amount.

For a typical agent with 2 simple tools:

| What | Approx. tokens |
|---|---|
| Tool schemas (2 tools) | ~300 |
| User message | ~30 |
| Each assistant tool-call request | ~50–80 |
| Each tool result | ~30–100 |

So input tokens grow roughly as:

| Iteration | Approx. cumulative input tokens |
|-----------|--------------------------------|
| 1 | ~330 (schemas + user message) |
| 2 | ~500 (+tool call + result) |
| 3 | ~670 (+another round) |
| 10 | ~1,500–2,000 |

Growth is linear, not exponential — but it still adds up when running thousands of agent jobs over a large dataset.

**Practical limits:**
- Cap iterations with .
- Truncate or summarize long tool outputs before appending them to message history.
- Use  for tool-heavy loops; reserve  for the final synthesis step.

## Multi-Agent Systems

So far each of our agents has been a single LLM operating alone. A **multi-agent system** chains multiple specialized agents together — each with a defined role — so their outputs feed into one another.

The classic pattern in research: a **Writer** agent drafts a report, and an **Adversarial** (Critic) agent reviews it for errors, weaknesses, and logical gaps. The agents iterate until the critic is satisfied.

```
User goal
    │
    ▼
Writer Agent  ──── draft ────►  Critic Agent
    ▲                               │
    │         revised draft         │ feedback
    └───────────────────────────────┘
    (loop until critic approves or max rounds reached)
```

This pattern mirrors how academic peer review works — and catches hallucinations that a single agent would miss.

## Writer + Adversarial Agent: Agent Handoff

The key design decision is **how agents hand off work**:

- The **Writer** receives a task and produces a draft. It knows nothing about the critic.
- The **Critic** receives the draft and produces structured feedback with an `approved` flag.
- The **Orchestrator** (your Python code) reads the flag, decides whether to loop, and passes feedback back to the Writer as context.

The agents themselves are just LLM calls with carefully designed system prompts. The intelligence of the system comes from the orchestration loop — which is just Python.

````python
import json, openai, os
from dotenv import load_dotenv

load_dotenv()
client = openai.OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# ── Agent system prompts ──────────────────────────────────────────────────────

WRITER_SYSTEM = """
You are an expert economics research writer.
Given a topic and optional reviewer feedback, write a concise 3-paragraph policy brief.
Be specific: cite mechanisms, name potential confounders, and state the empirical
strategy you would use to test your claim.
If feedback is provided, revise the draft to address every point raised.
"""

CRITIC_SYSTEM = """
You are an adversarial peer reviewer for economics research briefs.
Find logical gaps, unsupported claims, missing caveats, and potential endogeneity problems.

Respond in valid JSON with exactly this structure:
{
  "approved": false,
  "issues": ["issue 1", "issue 2"],
  "summary": "one-sentence verdict"
}
Set approved to true ONLY if the brief is ready to publish with no material changes.
"""


def call_writer(topic: str, feedback: str | None = None) -> str:
    user_msg = f"Topic: {topic}"
    if feedback:
        user_msg += f"\n\nReviewer feedback to address:\n{feedback}"
    resp = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": WRITER_SYSTEM},
            {"role": "user", "content": user_msg},
        ],
        temperature=0.7,
    )
    return resp.choices[0].message.content


def call_critic(draft: str) -> dict:
    resp = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": CRITIC_SYSTEM},
            {"role": "user", "content": f"Policy brief to review:\n\n{draft}"},
        ],
        temperature=0.3,
        response_format={"type": "json_object"},
    )
    return json.loads(resp.choices[0].message.content)


# ── Writer–Critic loop ────────────────────────────────────────────────────────


def writer_critic_loop(topic: str, max_rounds: int = 4) -> str:
    feedback, draft = None, None
    for round_num in range(1, max_rounds + 1):
        print(f"\n{'=' * 60}\nRound {round_num}: Writer drafting...")
        draft = call_writer(topic, feedback)
        print(draft[:300] + "...")

        print(f"Round {round_num}: Critic reviewing...")
        review = call_critic(draft)
        print(f"Approved: {review['approved']}  |  {review['summary']}")
        for i, issue in enumerate(review.get("issues", []), 1):
            print(f"  {i}. {issue}")

        if review["approved"]:
            print(f"\nApproved after {round_num} round(s).")
            break
        feedback = "\n".join(f"- {iss}" for iss in review["issues"])

    return draft


final_draft = writer_critic_loop(
    topic="The effect of minimum wage increases on employment in low-skill labor markets",
    max_rounds=3,
)
print("\n\n=== FINAL APPROVED DRAFT ===")
print(final_draft)
````

## Full Research Report Pipeline: Orchestrating Multiple Agents

A more realistic research workflow involves several specialized agents, each handling one stage:

```
User: "Write a report on X"
            │
            ▼
    ┌─────────────────┐   tool_choice="required"
    │ Literature Agent│──────────────────────────► search_semantic_scholar()
    │   (LLM)         │◄──────────────────────────  real papers from API
    └────────┬────────┘  synthesizes results
             │ literature context
             ▼
    ┌─────────────────┐
    │ Analysis Agent  │  — proposes empirical strategy
    └────────┬────────┘
             │ analysis memo
             ▼
    ┌─────────────────┐
    │  Writer Agent   │◄──────────────────────┐
    └────────┬────────┘                       │ revised draft
             │ draft                          │
             ▼                                │
    ┌─────────────────┐                       │
    │ Adversarial     │ ─── issues? ──────────┘
    │ Critic Agent    │
    └────────┬────────┘
             │ approved
             ▼
       Final Report
```

The **Literature Agent** uses tool calling with `tool_choice="required"` — the LLM must call `search_semantic_scholar` before it can answer. This is **Retrieval-Augmented Generation (RAG)**: the LLM formulates its own search query, we execute the real API call, and then the LLM synthesizes the actual results. It cannot hallucinate papers because it only summarizes what was retrieved.

The other three agents are plain LLM calls — tool calling is only needed where external data is required.

The orchestrator is just Python routing outputs between agents. This is the core pattern behind frameworks like LangGraph, CrewAI, and Anthropic's Claude Agents SDK.

````python
import json, textwrap, requests
from dataclasses import dataclass, field

# ── Shared helper ─────────────────────────────────────────────────────────────


def llm(
    system: str,
    user: str,
    temperature: float = 0.5,
    json_mode: bool = False,
    tools: list | None = None,
) -> str | dict:
    kwargs = dict(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": textwrap.dedent(system).strip()},
            {"role": "user", "content": user},
        ],
        temperature=temperature,
    )
    if json_mode:
        kwargs["response_format"] = {"type": "json_object"}
    if tools:
        kwargs["tools"] = tools
        kwargs["tool_choice"] = "auto"

    resp = client.chat.completions.create(**kwargs)
    msg = resp.choices[0].message

    # If the model decided to call a tool, return the raw message for the caller to handle
    if msg.tool_calls:
        return msg
    return msg.content


# ── Stage 1: Literature Agent ─────────────────────────────────────────────────


def search_semantic_scholar(query: str, n: int = 5) -> list[dict]:
    """Fetch real papers from the Semantic Scholar API."""
    url = "https://api.semanticscholar.org/graph/v1/paper/search"
    params = {
        "query": query,
        "limit": n,
        "fields": "title,authors,year,abstract,citationCount",
        "apikey": os.getenv("SEMANTIC_SCHOLAR_API_KEY"),
    }
    resp = requests.get(url, params=params, timeout=15)
    resp.raise_for_status()
    return resp.json().get("data", [])


def format_papers(papers: list[dict]) -> str:
    lines = []
    for i, p in enumerate(papers, 1):
        authors = ", ".join(a["name"] for a in p.get("authors", [])[:3])
        if len(p.get("authors", [])) > 3:
            authors += " et al."
        year = p.get("year", "n.d.")
        title = p.get("title", "Untitled")
        abstract = (p.get("abstract") or "No abstract available.")[
            :300
        ].rstrip() + "..."
        cites = p.get("citationCount", 0)
        lines.append(
            f"{i}. **{authors} ({year})**\n"
            f"   *{title}* [{cites} citations]\n"
            f"   {abstract}"
        )
    return "\n\n".join(lines)


LITERATURE_TOOLS = [
    {
        "type": "function",
        "function": {
            "name": "search_semantic_scholar",
            "description": (
                "Search the Semantic Scholar academic database for real published papers. "
                "Returns titles, authors, years, abstracts, and citation counts. "
                "Call this before summarizing the literature."
            ),
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "A precise academic search query, e.g. 'universal basic income labor supply RCT'",
                    },
                    "n": {
                        "type": "integer",
                        "description": "Number of papers to retrieve (default 5, max 10)",
                        "default": 5,
                    },
                },
                "required": ["query"],
            },
        },
    }
]

LITERATURE_TOOL_REGISTRY = {
    "search_semantic_scholar": lambda args: format_papers(
        search_semantic_scholar(args["query"], args.get("n", 5))
    ),
}


def literature_agent(topic: str) -> str:
    """
    Uses tool calling so the LLM decides what to search for.
    Pattern: LLM picks query → we execute the search → LLM synthesizes results.
    """
    print("  [Literature Agent] Deciding search strategy...")
    messages = [
        {
            "role": "system",
            "content": textwrap.dedent("""
                You are an economics literature search agent.
                Use the search_semantic_scholar tool to find relevant papers,
                then synthesize what the literature collectively says:
                key findings, identification strategies used, and open gaps.
                Reference each paper as Author (Year).
            """).strip(),
        },
        {"role": "user", "content": f"Find and summarize the literature on: {topic}"},
    ]

    # Round 1: let the LLM call the search tool
    resp = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        tools=LITERATURE_TOOLS,
        tool_choice="required",  # force it to search before answering
        temperature=0.3,
    )
    msg = resp.choices[0].message

    # Execute every tool call the model requested
    messages.append(msg)
    for tc in msg.tool_calls:
        args = json.loads(tc.function.arguments)
        print(f"  [Literature Agent] Searching: {args.get('query')!r}")
        result = LITERATURE_TOOL_REGISTRY[tc.function.name](args)
        messages.append({
            "role": "tool",
            "tool_call_id": tc.id,
            "content": result,
        })

    # Round 2: LLM synthesizes the real search results
    print("  [Literature Agent] Synthesizing results...")
    final = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=0.3,
    )
    return final.choices[0].message.content


# ── Stage 2: Analysis Agent ───────────────────────────────────────────────────


def analysis_agent(topic: str, literature: str) -> str:
    print("  [Analysis Agent] Designing empirical strategy...")
    return llm(
        system="""
        You are an applied econometrics specialist.
        Given a research topic and a literature summary, propose:
        1. The key causal question and null hypothesis.
        2. An identification strategy (IV, DiD, RDD, etc.) with justification.
        3. The primary outcome variable and at least one confounder to control for.
        4. A data source that could be used.
        Be concrete and technically precise.
        """,
        user=f"Topic: {topic}\n\nLiterature:\n{literature}",
        temperature=0.4,
    )


# ── Stage 3: Writer Agent ─────────────────────────────────────────────────────


def writer_agent(
    topic: str, literature: str, analysis: str, feedback: str | None = None
) -> str:
    print("  [Writer Agent] Drafting report...")
    user_content = (
        f"Topic: {topic}\n\n"
        f"Literature Summary:\n{literature}\n\n"
        f"Empirical Analysis Memo:\n{analysis}"
    )
    if feedback:
        user_content += f"\n\nRevisions requested by reviewer:\n{feedback}"
    return llm(
        system="""
        You are an economics research writer producing a policy brief.
        Synthesize the provided literature and analysis into a 4-paragraph report:
          1. Motivation and policy relevance.
          2. What the literature says (cite Author, Year).
          3. Proposed empirical strategy and expected findings.
          4. Policy implications and caveats.
        If reviewer feedback is provided, address every point in your revision.
        Be rigorous — vague claims will be rejected.
        """,
        user=user_content,
        temperature=0.6,
    )


# ── Stage 4: Adversarial Critic ───────────────────────────────────────────────


def critic_agent(draft: str) -> dict:
    print("  [Critic Agent] Reviewing draft...")
    raw = llm(
        system="""
        You are an adversarial peer reviewer for economics policy briefs.
        Identify: unsupported empirical claims, missing confounders, endogeneity
        concerns, vague identification strategies, and logical gaps.
        Return JSON:
        {
          "approved": false,
          "issues": ["...", "..."],
          "summary": "one-sentence verdict"
        }
        Set approved=true only if no material issues remain.
        """,
        user=f"Review this draft:\n\n{draft}",
        temperature=0.2,
        json_mode=True,
    )
    return json.loads(raw)


# ── Orchestrator ──────────────────────────────────────────────────────────────


@dataclass
class ReportState:
    topic: str
    literature: str = ""
    analysis: str = ""
    draft: str = ""
    review_history: list = field(default_factory=list)
    approved: bool = False


def run_research_pipeline(topic: str, max_revision_rounds: int = 3) -> ReportState:
    state = ReportState(topic=topic)

    print("\n[Pipeline] Stage 1: Literature")
    state.literature = literature_agent(topic)

    print("\n[Pipeline] Stage 2: Analysis")
    state.analysis = analysis_agent(topic, state.literature)

    feedback = None
    for round_num in range(1, max_revision_rounds + 1):
        print(f"\n[Pipeline] Stage 3: Writing (round {round_num})")
        state.draft = writer_agent(topic, state.literature, state.analysis, feedback)

        print(f"\n[Pipeline] Stage 4: Critique (round {round_num})")
        review = critic_agent(state.draft)
        state.review_history.append(review)

        print(f"  Approved: {review['approved']} | {review['summary']}")
        for i, issue in enumerate(review.get("issues", []), 1):
            print(f"    {i}. {issue}")

        if review["approved"]:
            state.approved = True
            print(f"\n[Pipeline] Report approved after {round_num} revision round(s).")
            break

        feedback = "\n".join(f"- {iss}" for iss in review["issues"])

    if not state.approved:
        print("\n[Pipeline] Max rounds reached. Returning best draft.")

    return state


# ── Run the pipeline ──────────────────────────────────────────────────────────
state = run_research_pipeline(
    topic="Effects of universal basic income on labor supply and earnings",
    max_revision_rounds=3,
)

print("\n" + "=" * 70)
print("FINAL REPORT")
print("=" * 70)
print(state.draft)
````

## What the Pipeline Teaches Us

**Separation of concerns** — each agent only needs to be good at one narrow task. This is easier to prompt, debug, and improve than a single "do everything" agent.

**State as the glue** — the `ReportState` dataclass is the shared memory between agents. In production systems, this might be a database row, a shared file, or a message queue.

**The critic is the most important agent** — without adversarial review, the writer will produce confident-sounding but potentially wrong output. The critic forces the system to iterate toward quality.

**When to use multi-agent systems:**
- Tasks with clearly separable stages (search → analyze → write → review)
- When different stages benefit from different temperatures or models
- When you want humans to be able to inspect or intervene at each stage

**When not to:**
- Simple single-step tasks — overhead is not worth it
- When you need low latency — each agent call adds round-trip time
- When the stages are tightly coupled — passing context between agents becomes expensive

## More Control: AutoTokenizer + AutoModelForCausalLM

For full control over tokenization, generation parameters, and batching, load the tokenizer and model separately.

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_name = "HuggingFaceTB/SmolLM2-360M-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype=torch.float16,  # half precision — 2× memory savings vs float32
    device_map="auto",
)

prompt = "Explain the Fisher equation in one paragraph."
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

with torch.no_grad():
    output_ids = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.7,
        do_sample=True,
        repetition_penalty=1.1,
    )

# Decode only the newly generated tokens (skip the prompt)
new_tokens = output_ids[0][inputs["input_ids"].shape[1] :]
print(tokenizer.decode(new_tokens, skip_special_tokens=True))
````

## Key Generation Parameters

These are the same conceptual knobs as the OpenAI API — the parameter names just differ slightly.

| Parameter | What it controls | Typical range |
|---|---|---|
| `max_new_tokens` | Max tokens to generate | 64–4096 |
| `temperature` | Randomness of sampling | 0.1 (focused) – 1.5 (creative) |
| `top_p` | Nucleus sampling cutoff | 0.85–0.95 |
| `top_k` | Restrict to top-k tokens at each step | 20–100 |
| `repetition_penalty` | Penalize repeating the same text | 1.0 (none) – 1.3 (strong) |
| `do_sample` | If False: greedy decoding (deterministic) | True/False |

**Rule of thumb:** For factual data extraction tasks, use low temperature (0.1–0.3) and `do_sample=False`. For creative or open-ended tasks, use temperature 0.7–1.0.

## Quantization: Fitting Large Models on Consumer Hardware

A 7B-parameter model in 32-bit floats requires ~28 GB of VRAM — more than most GPUs have.

**Quantization** reduces precision to trade a small accuracy loss for a dramatic memory reduction:

| Precision | Memory for 7B model | Quality |
|---|---|---|
| float32 (default) | ~28 GB | Baseline |
| bfloat16 | ~14 GB | Negligible loss |
| int8 | ~7 GB | Small loss |
| int4 (4-bit) | ~3.5 GB | Moderate loss, still useful |

The `bitsandbytes` library enables 4-bit and 8-bit quantization with one config object. `GPTQ` and `AWQ` are alternative quantization formats you'll find on the Hub — they bake quantization into the saved weights, so no extra library is needed at load time.
